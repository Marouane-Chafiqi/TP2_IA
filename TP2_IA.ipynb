```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

phrases = [
    "j aime le football et le sport",
    "j aime le cinema et les films",
    "le tennis est un sport rapide"
]

bow = CountVectorizer()
matrice = bow.fit_transform(phrases)

print("Mots extraits :", bow.get_feature_names_out())
print(matrice.toarray())
```

```python
tableau_bow = pd.DataFrame(matrice.toarray(), columns=bow.get_feature_names_out())
print(tableau_bow)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
matrice_tfidf = tfidf.fit_transform(phrases)

tableau_tfidf = pd.DataFrame(matrice_tfidf.toarray(), columns=tfidf.get_feature_names_out())
print(tableau_tfidf.round(2))
```

```python
for i in range(len(phrases)):
    ligne = tableau_tfidf.iloc[i]
    print("Phrase", i + 1, "-> mot le plus important :", ligne.idxmax())
```
